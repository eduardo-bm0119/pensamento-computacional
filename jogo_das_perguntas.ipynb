```python
#jogo da memória

import random
import time
from IPython.display import clear_output

def jogar_jogo_da_memoria():
    nivel = 1
    tamanho_sequencia = 5
    tempo_exibicao = 3.0

    print("=== BEM-VINDO AO JOGO DA MEMÓRIA ===\n")
    input("Pressione ENTER para começar o jogo...")

    while True:
        # Limpa tudo e avisa o início do nível
        clear_output(wait=False)
        print(f"--- NÍVEL {nivel} ---")
        print(f"Prepare-se! Você terá {tempo_exibicao:.1f} segundos para memorizar a sequência.")
        time.sleep(1.5)

        # Gerar a sequência de números aleatórios
        sequencia = [str(random.randint(0, 9)) for _ in range(tamanho_sequencia)]
        sequencia_str = " ".join(sequencia)

        # Mostrar a sequência e limpar imediatamente antes para garantir foco
        clear_output(wait=False)
        print(f"--- NÍVEL {nivel} ---")
        print("\nMEMORIZE ESTA SEQUÊNCIA:\n")
        print(f"👉  {sequencia_str}  👈")
        print("\n-------------------------")

        # Aguardar o tempo limite com os números visíveis
        time.sleep(tempo_exibicao)

        # IMPORTANTE: Limpar a tela imediatamente (wait=False) para sumir com os números
        clear_output(wait=False)
        time.sleep(0.1)

        # Só agora pedimos a resposta do usuário
        print(f"--- NÍVEL {nivel} ---")
        print("O tempo acabou! Digite a sequência que estava na tela.")
        resposta_usuario = input("Sua resposta (pode ser com ou sem espaços): ")

        # Tratar a resposta (remover espaços e comparar)
        resposta_limpa = resposta_usuario.replace(" ", "")
        sequencia_limpa = "".join(sequencia)

        if resposta_limpa == sequencia_limpa:
            print("\n🎉 PARABÉNS! Você acertou! 🎉")
            nivel += 1
            tamanho_sequencia += 1
            tempo_exibicao += 0.5
            time.sleep(2)
        else:
            print(f"\n❌ QUE PENA! Você errou. ❌")
            print(f"A sequência correta era: {sequencia_str}")
            print(f"Você parou no Nível {nivel}.")

            jogar_novamente = input("\nDeseja jogar novamente? (s/n): ").strip().lower()
            if jogar_novamente == 's':
                nivel = 1
                tamanho_sequencia = 5
                tempo_exibicao = 3.0
            else:
                print("\nObrigado por jogar! Até a próxima.")
                break

# Iniciar o jogo
jogar_jogo_da_memoria()
```
